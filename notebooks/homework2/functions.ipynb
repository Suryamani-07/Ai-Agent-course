```python
from agents import Agent, Runner, trace , function_tool
from openai.types.responses import ResponseTextDeltaEvent
```

```python
from temperature_tool import convert_temperature
from discount_tool import calculate_discount
```

```python
utility_agent = Agent(
    name="Utility Assistant",
    instructions="""
    You are a utility assistant.
    Use the available tools to convert temperatures
    and calculate discounted prices.
    """,
    tools=[convert_temperature, calculate_discount]
)
```

```python
with trace("Utility Assistant"):
    result = await Runner.run(
        utility_agent,
        "Convert 30 Celsius to Fahrenheit and calculate the price of ₹2000 after a 15% discount."
    )

print(result.final_output)
```
