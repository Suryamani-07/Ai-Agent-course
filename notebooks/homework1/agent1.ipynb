```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
resume_agent = Agent(
    name="Resume Review Assistant",
    instructions="""
    You are a professional resume reviewer.

    - Review resume content.
    - Identify grammar and spelling mistakes.
    - Suggest improvements to skills and experience sections.
    - Recommend ways to make the resume ATS-friendly.
    - Give clear and practical feedback.
    """
)
```

```python
with trace("Resume Review Agent"):
    result = await Runner.run(
        resume_agent,
        "Suggest improvements for a fresher's software developer resume."
    )

print(result.final_output)
```
