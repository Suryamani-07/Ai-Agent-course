```python
import dotenv
import os

from openai import OpenAI

dotenv.load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    print(
        """Error: OPENAI_API_KEY environment variable not set. Please copy the .env.template file as .env and fill it in.
    
    You can execute these commands in the terminal to get started:
    cp .env.template .env
    code .env
    """
    )

# Test OpenAI Access
print(
    OpenAI()
    .responses.create(
        model=os.environ["OPENAI_DEFAULT_MODEL"], input="Say: We are up and running!"
    )
    .output_text
)
```

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
quiz_agent = Agent(
    name="Quiz Master",
    instructions="""
    You are an educational quiz assistant.

    - Generate multiple-choice questions on requested topics.
    - Provide four options for each question.
    - Check the user's answers.
    - Explain why an answer is correct or incorrect.
    - Keep questions clear and educational.
    """
)
```

```python
with trace("Quiz Master Agent"):
    result = await Runner.run(
        quiz_agent,
        "Create 5 multiple-choice questions about Python programming with answers."
    )

print(result.final_output)
```
